# **Text Processing**
---
The text processing phase is a preliminary stage where the text to be manipulated is processed to be ready for subsequent analysis.

Text processing usually entails several steps that could possibly include:
- **Language Identification**: identifying the language of a given text.
- **Tokenization**: splitting a given text into several sentences/words.
- **Dependency tree parsing:** analyzing the dependencies between words composing the text.
- **Stemming/Lemmatization:** obtain the root form for each word in text.
- **Stopword removal**: removing words that are commonly used that they carry very little useful information.
- **Part of Speech Tagging:** given a word, retrieve its part of speech (proper noun, common noun or verb).

```python
import langid
import pandas as pd
from iso639 import Lang
from sklearn.metrics import accuracy_score
import time
import langdetect
from langdetect.lang_detect_exception import LangDetectException # Importa l'eccezione

df = pd.read_csv("langid_dataset.csv", sep=",")
df.head()
df['language'] = df['language'].apply(lambda x: Lang(x.strip().title()).pt1)
start_time= time.time()
df['lingua_predetta_langid'] = df['Text'].apply(lambda x: langid.classify(x)[0])
end_time= time.time()
tempo_tot= end_time- start_time
num_es= len(df)
tempo_medio_langid= tempo_tot/ num_es
def safe_langdetect(testo):
    testo_str = str(testo).strip()
    if not testo_str:
        return 'und'
    try:
        # Ritorna direttamente la stringa, SENZA [0]
        return langdetect.detect(testo_str)
    except LangDetectException:
        # Se il testo contiene solo numeri o simboli, langdetect fallisce.
        # Catturiamo l'errore e restituiamo 'und' (undefined)
        return 'und'
start_time_lang_detect= time.time()
df['lingua_predetta_langdetect'] = df['Text'].apply(safe_langdetect)
end_time_langdetect= time.time()
tempo_tot_detect= end_time_langdetect- start_time_lang_detect
tempo_medio_langdetect= tempo_tot_detect/ num_es
```

```python
import fasttext
import os
import urllib.request

model_path = 'lid.176.ftz'
url = 'https://dl.fbaipublicfiles.com/fasttext/supervised-models/lid.176.ftz'
urllib.request.urlretrieve(url, model_path)
model= fasttext.load_model(model_path)
def safe_fasttext(testo):
    # FastText va in errore se il testo contiene ritorni a capo (\n)
    testo_str = str(testo).strip().replace('\n', ' ')
    if not testo_str:
        return 'und'
    try:
        # model.predict restituisce qualcosa come: (('__label__it',), array([0.99]))
        predizione = model.predict(testo_str)[0][0]
        # Rimuoviamo '__label__' per avere solo 'it', 'en', ecc. e confrontarlo con gli altri
        return predizione.replace('__label__', '')
    except Exception:
        return 'und'
start_time_fasttext = time.time()
df['lingua_predetta_fasttext'] = df['Text'].apply(safe_fasttext)
end_time_fasttext = time.time()
tempo_tot_fasttext = end_time_fasttext - start_time_fasttext
num_es = len(df)
tempo_medio_fasttext = tempo_tot_fasttext / num_es
```

```python
accuracy_langid = accuracy_score(df['language'], df['lingua_predetta_langid'])
accuracy_langdetect = accuracy_score(df['language'], df['lingua_predetta_langdetect'])
accuracy_ft = accuracy_score(df['language'], df['lingua_predetta_fasttext'])

print(f"\nLangid\n")
print(f"\nAccuratezza complessiva:{accuracy_langid *100:.2f}%\n")
print(f"\ntempo medio: {tempo_medio_langid:.4f}\n")
print(f"\nLangdetect\n")
print(f"\nAccuratezza complessiva:{accuracy_langdetect *100:.2f}%\n")
print(f"\ntempo medio: {tempo_medio_langdetect:.4f}\n")
print(f"\nFasttext\n")
print(f"\nAccuratezza complessiva:{accuracy_ft *100:.2f}%\n")
print(f"\ntempo medio: {tempo_medio_fasttext:.4f}\n")
```

For English-written text, apply word-level tokenization. What is the average number of words per sentence?

Implement word-tokenization using both [nltk](https://www.nltk.org/) and [spacy](https://spacy.io/). Report the results for both of them.

For spaCy use the `en_core_web_sm` model.

```python
# your code here
import nltk
from nltk.tokenize import sent_tokenize, word_tokenize

# Scarica i modelli necessari per la tokenizzazione se non li hai già
nltk.download('punkt_tab')

# 1. Filtra il DataFrame per mantenere solo i testi in inglese
# (usiamo la colonna della ground truth 'language')
df_en = df[df['language'] == 'en'].dropna(subset=['Text'])

def conta_metriche(testo):
    testo_str = str(testo).strip()
    if not testo_str:
        return 0, 0

    # Tokenizzazione delle frasi (sentence-level)
    frasi = sent_tokenize(testo_str)
    numero_frasi = len(frasi)

    numero_parole = 0
    for frase in frasi:
        # Tokenizzazione delle parole (word-level) per ogni singola frase
        parole = word_tokenize(frase)

        # word_tokenize considera anche la punteggiatura come token (es. ",", ".").
        # Filtriamo mantenendo solo le vere e proprie parole alfanumeriche.
        parole_vere = [p for p in parole if p.isalnum()]
        numero_parole += len(parole_vere)

    return numero_parole, numero_frasi

# 2. Applica la funzione al testo in inglese
# Restituisce una Series di tuple: (numero_parole, numero_frasi)
conteggi = df_en['Text'].apply(conta_metriche)

# 3. Somma totale di parole e frasi
totale_parole = sum(c[0] for c in conteggi)
totale_frasi = sum(c[1] for c in conteggi)

# 4. Calcolo della media
if totale_frasi > 0:
    media_parole_per_frase = totale_parole / totale_frasi
    print(f"Testi in inglese analizzati: {len(df_en)}")
    print(f"Totale frasi: {totale_frasi}")
    print(f"Totale parole valide (esclusa punteggiatura): {totale_parole}")
    print(f"\nMedia esatta: {media_parole_per_frase:.2f} parole per frase")
else:
    print("Nessun testo in inglese trovato nel dataset.")
```

```python
import spacy
import pandas as pd

# 1. Carica il modello linguistico inglese di spaCy
nlp = spacy.load("en_core_web_sm")

# 2. Filtra il DataFrame per mantenere solo i testi in inglese
df_en = df[df['language'] == 'en'].dropna(subset=['Text'])

totale_parole = 0
totale_frasi = 0

# 3. Processa i testi in modo efficiente usando nlp.pipe
# (Disabilitiamo 'ner' e 'lemmatizer' perché per contare le parole non ci servono,
# rendendo l'esecuzione ancora più veloce)
for doc in nlp.pipe(df_en['Text'].astype(str), disable=["ner", "lemmatizer"]):

    # doc.sents divide automaticamente il testo in frasi
    for frase in doc.sents:
        totale_frasi += 1

        # Iteriamo sui token della frase. Usiamo token.text.isalnum() per coerenza
        # assoluta con il filtro usato nell'esempio NLTK (escludendo la punteggiatura).
        # In alternativa, spaCy offre anche il comodo attributo nativo: non token.is_punct
        parole_vere = [token for token in frase if token.text.isalnum()]

        totale_parole += len(parole_vere)

# 4. Calcolo e stampa della media
if totale_frasi > 0:
    media_parole_per_frase = totale_parole / totale_frasi
    print(f"Testi in inglese analizzati: {len(df_en)}")
    print(f"Totale frasi: {totale_frasi}")
    print(f"Totale parole valide (esclusa punteggiatura): {totale_parole}")
    print(f"\nMedia esatta con spaCy: {media_parole_per_frase:.2f} parole per frase")
else:
    print("Nessun testo in inglese trovato nel dataset.")
```

## **Occurrence-based text representation - TF-IDF**

TF-IDF (term frequency-inverse document frequency) is a statistical measure that evaluates how relevant a word is to a document in a collection of documents. It allows to create occurrence-based vector representation for each document.

## Exercise 5
Use TF-IDF to vectorize each sentence in the original data collection. You can choose your preferred implementation for TF-IDF vectorization. It is also available on [SciKit-Learn library](https://scikit-learn.org/stable/modules/generated/sklearn.feature_extraction.text.TfidfVectorizer.html)

```python
# your code here
from sklearn.feature_extraction.text import TfidfVectorizer
import pandas as pd

# 1. Pulizia preventiva: assicuriamoci di non avere valori nulli o non-stringa,
# altrimenti il vettorizzatore andrà in crash
df_clean = df.dropna(subset=['Text']).copy()
testi = df_clean['Text'].astype(str)

# 2. Inizializza il vettorizzatore TF-IDF
# Puoi aggiungere parametri come stop_words='english' o max_features=5000
# per limitare il vocabolario ed eliminare parole troppo comuni, ma qui
# manteniamo le impostazioni di base
vectorizer = TfidfVectorizer()

# 3. Calcola il TF-IDF e trasforma il testo
# fit_transform() impara il vocabolario e restituisce direttamente la matrice vettorizzata
tfidf_matrix = vectorizer.fit_transform(testi)

# 4. Stampa le informazioni sulla matrice risultante
print(f"Dimensioni della matrice TF-IDF: {tfidf_matrix.shape}")
print(f"Numero di documenti (frasi): {tfidf_matrix.shape[0]}")
print(f"Numero di feature (parole uniche nel vocabolario): {tfidf_matrix.shape[1]}\n")

# --- VISUALIZZAZIONE DEL RISULTATO ---
# La tfidf_matrix è una "matrice sparsa" (scipy.sparse) per risparmiare RAM.
# Per visualizzarne una parte, convertiamo solo le prime 5 righe in un DataFrame.

# Estraiamo i nomi delle colonne (cioè le parole reali del vocabolario)
feature_names = vectorizer.get_feature_names_out()

# Creiamo un DataFrame con le prime 5 righe
df_tfidf_preview = pd.DataFrame(tfidf_matrix[:5].toarray(), columns=feature_names)

# Poiché il dataframe avrebbe decine di migliaia di colonne piene di zeri,
# filtriamo per mostrare SOLO le colonne che hanno un valore > 0 in queste prime 5 righe
colonne_attive = df_tfidf_preview.columns[(df_tfidf_preview > 0).any()]

print("Anteprima dei punteggi TF-IDF per le prime 5 frasi:")
print(df_tfidf_preview[colonne_attive])
```

## Exercise 6

Build a supervised multi-class language detector using as features the vector obtained by TF-IDF representation. Use 80% of the data to train the language detector and 20% of the data for assessing its accuracy.

```python
# your code here
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score, classification_report

# 1. Pulizia preventiva: assicuriamoci di non avere righe con testo o etichetta mancante
df_clean = df.dropna(subset=['Text', 'language']).copy()

# Variabile indipendente (X) e variabile dipendente/target (y)
X = df_clean['Text'].astype(str)
y = df_clean['language']

# 2. Suddivisione del dataset (80% Train, 20% Test)
# Utilizziamo stratify=y per assicurarci che la proporzione di ogni lingua
# sia identica sia nel set di addestramento che in quello di test
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42, stratify=y
)

print(f"Dimensione Training set: {X_train.shape[0]} frasi")
print(f"Dimensione Test set: {X_test.shape[0]} frasi\n")

# 3. Vettorizzazione TF-IDF
vectorizer = TfidfVectorizer()

# ATTENZIONE: Usiamo fit_transform() SUL TRAIN SET per creare il vocabolario
X_train_tfidf = vectorizer.fit_transform(X_train)

# Usiamo SOLO transform() SUL TEST SET.
# Se usassimo fit_transform anche qui, il modello imparerebbe parole "nuove" dal test set,
# commettendo il grave errore di "data leakage" (sbirciare le risposte prima dell'esame).
X_test_tfidf = vectorizer.transform(X_test)

# 4. Inizializzazione e addestramento del classificatore
clf = MultinomialNB()
clf.fit(X_train_tfidf, y_train)

# 5. Predizione sul test set (il 20% dei dati mai visti prima dal modello)
y_pred = clf.predict(X_test_tfidf)

# 6. Valutazione delle performance
accuracy = accuracy_score(y_test, y_pred)

print(f"--- RISULTATI DEL MODELLO ---")
print(f"Accuratezza complessiva: {accuracy * 100:.2f}%\n")
print(classification_report(y_test, y_pred))
```

```python
# your code here
import pandas as pd
from gensim import corpora, models
from gensim.utils import simple_preprocess
from gensim.parsing.preprocessing import STOPWORDS

# Load the dataset (Make sure you have run the !wget command first)
df = pd.read_csv('CovidFake_filtered.csv')

# 1. Create a corpus composed of the headlines
# Note: Assuming the headline column is named 'title'. Change to 'headlines' or 'text' if different in the CSV.
headlines = df['headlines'].dropna().tolist()

# Tokenize the documents (lowercases, removes punctuation, and splits into words)
tokenized_headlines = []
for headline in headlines:
    # Tokenizza, mette in minuscolo e tiene solo le parole che NON sono nelle STOPWORDS
    tokens = [word for word in simple_preprocess(headline) if word not in STOPWORDS]
    tokenized_headlines.append(tokens)

# 2. Generate a dictionary to create a word -> id mapping
dictionary = corpora.Dictionary(tokenized_headlines)

# Optional but recommended: Filter out extreme frequencies to improve topics
dictionary.filter_extremes(no_below=5, no_above=0.5)

# 3. Preprocess the corpus to obtain the representation required for LSI
# LSI requires a Bag-of-Words (BoW) representation: a list of (word_id, word_frequency) tuples
bow_corpus = [dictionary.doc2bow(text) for text in tokenized_headlines]

# Alternatively, LSI often performs better when applied over a TF-IDF weighted corpus:
# tfidf_model = models.TfidfModel(bow_corpus)
# tfidf_corpus = tfidf_model[bow_corpus]

# 4. Train the LSI model and inspect the top-5 topics generated
# (Using bow_corpus here. Swap with tfidf_corpus if you applied TF-IDF weights)
lsi_model = models.LsiModel(bow_corpus, id2word=dictionary, num_topics=5)

# Inspect the top-5 topics
print("Top 5 LSI Topics:")
top_topics = lsi_model.print_topics(num_topics=5, num_words=10)
for topic in top_topics:
    print(topic)
```

This function `generate_topics` is designed to make it easy for you to explore
how a language model extracts topics from sentences.  

- You do **not** need to modify the function itself.
- In the upcoming steps, all you will do is **call this function** with a new
  prompt describing the topics you want the model to extract.
- The function will:
    1. Randomly pick a few sentences from the dataset.
    2. Send them to the model with your prompt.
    3. Print each sentence along with the predicted topics.
- Make sure your prompt clearly instructs the model on what format to return.

```python
def generate_topics(prompt, model, tokenizer, sentences, n_examples=5, device='cuda'):
    """
    Generate topics for random sentences using a language model.

    Args:
        prompt (str): Instruction for the model.
        model: The language model.
        tokenizer: The tokenizer for the model.
        sentences (list of str): Sentences to select from.
        n_examples (int): Number of sentences to process.
        device (str): Device to run the model on.

    Returns:
        list of (sentence, topics) tuples.
    """
    results = []
    for i, sentence in enumerate(random.choices(sentences, k=n_examples)):
        full_prompt = f"{prompt}{sentence}\nTopics: "
        inputs = tokenizer(full_prompt, return_tensors="pt").to(device)

        output_ids = model.generate(**inputs, max_new_tokens=32)
        decoded = tokenizer.batch_decode(output_ids, skip_special_tokens=True)[0]

        if "Topics:" in decoded:
            topics = decoded.split("Topics:")[-1].strip()
        else:
            topics = decoded.strip()

        results.append((sentence, topics))
        print(f"Sentence {i+1}: {sentence}\nTopics: {topics}\n{'-'*50}")
    return results
```


**1<sup>st</sup> approach**: Ask the model to identify the topic(s) contained in a given sentence <u>without providing</u> a predefined list of topics to choose from.\
*Example of prompt*:\
Which are the most relevant topics of the following sentence?

```python
PROMPT1 = """Which are the most relevant topics of the following sentence?
Provide only a concise, comma-separated list of 1 to 3 keywords. Do not write full sentences or explanations.

Sentence: """
generate_topics(PROMPT1, model, tokenizer, sentences, n_examples=10, device=device)
```

**2<sup>nd</sup> approach**: Ask the model to identify the topic(s) contained in a given sentence <u>providing</u> a predefined list of topics to choose from.

*Example of prompt*:\
Which are the most relevant topics of the following sentence?\
Choose among: medicine, COVID, Artificial Intelligence, treatment, English literature, vaccine, gardening

```python
PROMPT2 = """Which are the most relevant topics of the following sentence?
Choose ONLY from the following list: medicine, COVID, Artificial Intelligence, treatment, English literature, vaccine, gardening.
Provide your answer as a comma-separated list of the selected topics. Do not write any other text or explanations.

Sentence: """
generate_topics(PROMPT2, model, tokenizer, sentences, n_examples=10, device=device)
```

**3<sup>rd</sup> approach**: Ask the model to identify the topic(s) contained in a given sentence <u>providing</u> a predefined list of topics to choose from along with the corresponding definitions.

*Example of prompt*:\
Which are the most relevant topics of the following sentence?\
Choose among:
- medicine: treatment for illness or injury, or the study of this
- COVID: an infectious disease caused by a coronavirus
- Artificial Intelligence: computer systems that have some of the qualities that the human brain has, such as learn from data
- treatment: the use of drugs to cure a person of an illness or injury
- English literature: artistic works written in the English language, especially those with a high and lasting artistic value
- vaccine: a substance that is put into the body of a person or animal to protect them from a disease
- gardening: the job or activity of working in a garden

Definitions are taken from the [Cambridge Dictionary](https://dictionary.cambridge.org/) and have been slightly adapted.

```python
PROMPT3 = """Which are the most relevant topics of the following sentence?
Choose ONLY among the following topics:
- medicine: treatment for illness or injury, or the study of this
- COVID: an infectious disease caused by a coronavirus
- Artificial Intelligence: computer systems that have some of the qualities that the human brain has, such as learn from data
- treatment: the use of drugs to cure a person of an illness or injury
- English literature: artistic works written in the English language, especially those with a high and lasting artistic value
- vaccine: a substance that is put into the body of a person or animal to protect them from a disease
- gardening: the job or activity of working in a garden

Provide your answer as a comma-separated list of the selected topics. Do not write any other text.

Sentence: """
generate_topics(PROMPT3, model, tokenizer, sentences, n_examples=10, device=device)
```
